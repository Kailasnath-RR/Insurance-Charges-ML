##### Observation:

## RELATIONSHIP BETWEEN SEX,SMOKER AND INSURANCE CHARGES:

```python
ins_data.groupby(['sex','smoker'])['charges'].mean()
```

- The average insurance charges for smokers are significantly higher than for non-smokers, regardless of sex.
- This indicates that smoking is a major factor influencing insurance charges.

## RELATIONSHIP BETWEEN AGE,SMOKER AND INSURANCE CHARGES:

```python
sns.scatterplot(y=ins_data['charges'],x=ins_data['age'],hue=ins_data['smoker'])
```

- The scatter plot shows a clear positive correlation between age and insurance charges, indicating that charges tend to increase as age increases.
- Additionally, smoking status has a significant impact on insurance charges. 
- Smokers consistently exhibit much higher charges compared to non-smokers across all age groups, highlighting smoking as a major factor influencing insurance costs.

```python
sns.boxplot(x=ins_data['smoker'],y=ins_data['charges'])
```

### RELATIONSHIP BETWEEN BMI , CHARGES AND SMOKER :

```python
sns.scatterplot(x=ins_data['bmi'],y=ins_data['charges'],hue = ins_data['smoker'])
plt.title("BMI vs Insurance Charges")
```

The scatter plot shows that insurance charges tend to increase slightly with increasing BMI, indicating a weak to moderate positive relationship between BMI and charges. However, smoking status has a much stronger impact on insurance charges than BMI alone.

##### Smokers consistently incur significantly higher charges across all BMI ranges.

#### Implications:
- Both BMI and smoking status are important features for predicting insurance charges. However, smoking status appears to be a significantly stronger predictor compared to BMI.
- There could be a combined effect of BMI and Smoking on insurance charges.
